```python
# Install dependencies
# %pip install python-dotenv typesafe_sdk
from dotenv import load_dotenv
import os
load_dotenv()
from typesafe_sdk import TypeSafeClient
MODEL_NAME="jev-latest"
client = TypeSafeClient(
    api_key=os.environ["TYPESAFE_API_KEY"],
    base_url=os.environ["TYPESAFE_BASE_URL"],
)
```

```python
states = [
    "I was charged twice for my subscription. Please refund the duplicate charge.",
    "I forgot to cancel my subscription before it renewed. Can I get my money back?",
    "I was charged twice, but I’m just letting you know—no refund is needed.",
    "How do I check what I was charged for?",
    "I’m unhappy with the subscription price and might cancel.",
]

for state in states:
    result = client.system_one(
        model=MODEL_NAME,
        state=state,
        questions={
            "refund": {
                "type": "noul",
                "instructions": (
                    "Return a number from 0 to 1 indicating how strongly the "
                    "customer is asking for a refund. Use 0 for no refund request "
                    "and 1 for an explicit refund request."
                ),
            },
        },
    )
    print({"state": state, "result": result})
```
